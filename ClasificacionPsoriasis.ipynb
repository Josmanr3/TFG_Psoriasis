```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import KFold, StratifiedKFold, train_test_split
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.applications.efficientnet import preprocess_input
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.models import Model
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, f1_score
from tensorflow.keras import models, regularizers
from collections import Counter
import random

# --- Configuración General ---
DATASET_PATH = '/kaggle/input/psoriasis-ds/dataset_v8/dataset_v8'
IMG_SIZE = 224
BATCH_SIZE = 16
EPOCHS_KFOLD = 50
RANDOM_SEED = 44

# --- Configuración NUEVA: Split de Test ---
TEST_SPLIT_PERCENTAGE = 0.15 
PERFORM_FINE_TUNING = False  

# --- Configuración Específica para K-Fold Cross-Validation ---
N_SPLITS = 5
USE_STRATIFIED_KFOLD = True

# --- Funciones de Preprocesamiento y Balanceo ---

def load_all_images(directory, img_size):
    """
    Carga todas las imágenes del directorio dado, las preprocesa y devuelve los arrays
    de imágenes, etiquetas y las rutas de archivo originales.
    """
    classes = sorted(os.listdir(directory))
    images, labels, original_filepaths = [], [], []
    class_indices = {class_name: i for i, class_name in enumerate(classes)}

    class_counts = {class_name: 0 for class_name in classes}

    print(f"Cargando y redimensionando imágenes a {img_size}x{img_size}...")
    for class_name in classes:
        class_path = os.path.join(directory, class_name)
        if not os.path.isdir(class_path):
            continue

        for img_name in os.listdir(class_path):
            if img_name.lower().endswith(('.png', '.jpg', '.jpeg')):
                img_path = os.path.join(class_path, img_name)
                try:
                    img = tf.keras.preprocessing.image.load_img(img_path, target_size=(img_size, img_size))
                    img_array = tf.keras.preprocessing.image.img_to_array(img)
                    img_array = preprocess_input(img_array)
                    images.append(img_array)
                    labels.append(class_indices[class_name])
                    original_filepaths.append(img_path)
                    class_counts[class_name] += 1
                except Exception as e:
                    print(f"Error cargando imagen {img_path}: {e}")

    print("Distribución de clases cargadas (total del dataset):")
    for class_name, count in class_counts.items():
        print(f"    - {class_name}: {count} imágenes")

    return np.array(images), np.array(labels), classes, np.array(original_filepaths)

def balance_dataset_with_augmentation(X_data, y_data, class_names_list, target_img_size):
    """
    Balancea el dataset de entrenamiento usando técnicas de data augmentation.
    Asegura que el número de imágenes de cada clase llegue al doble de la clase mayoritaria.
    """
    print(f"Balanceando {len(X_data)} muestras de entrenamiento con aumentación avanzada...")

    datagen = ImageDataGenerator(
        rotation_range=30,
        width_shift_range=0.2,
        height_shift_range=0.2,
        shear_range=0.2,
        zoom_range=0.3,
        horizontal_flip=True,
        brightness_range=[0.9, 1.1],
        #channel_shift_range=30.0,
        fill_mode='nearest'
    )

    class_counts = Counter(y_data)
    if not class_counts:
        print("Advertencia: No hay datos para balancear.")
        return np.array([]), tf.keras.utils.to_categorical(np.array([]), num_classes=len(class_names_list))

    # Calcula la clase mayoritaria en el set de datos actual
    max_count_current_data = max(class_counts.values()) if class_counts else 0
    
    # Define el objetivo de balanceo 
    target_overall_count = int(max_count_current_data * 2.5)
    if target_overall_count == 0: # Evitar división por cero si max_count_current_data es 0
        target_overall_count = 1

    print(f"Clase mayoritaria actual en el set de entrenamiento: {max_count_current_data} imágenes. Objetivo de balanceo por clase: {target_overall_count} imágenes.")

    X_balanced, y_balanced = list(X_data), list(y_data)

    for class_idx in range(len(class_names_list)):
        current_indices = np.where(y_data == class_idx)[0]
        current_count = len(current_indices)
        
        if current_count < target_overall_count:
            needed = target_overall_count - current_count
            print(f"Generando {needed} imágenes adicionales para clase {class_names_list[class_idx]} (actual: {current_count})")
            images_to_augment = X_data[current_indices]

            if len(images_to_augment) == 0:
                print(f"Advertencia: No hay imágenes base para aumentar para la clase {class_names_list[class_idx]}.")
                continue

            # Calcular cuántas veces necesitamos repetir cada imagen original
            # para generar la cantidad necesaria
            repetitions_per_original_image = int(np.ceil(needed / len(images_to_augment)))
            
            augmented_images_for_class = []
            
            # Generar imágenes aumentadas hasta alcanzar la cantidad necesaria
            for _ in range(repetitions_per_original_image):
                if len(augmented_images_for_class) >= needed:
                    break
                    
                for img_original in images_to_augment:
                    if len(augmented_images_for_class) >= needed:
                        break
                        
                    # Generar 1 imagen aumentada por iteración de la imagen original
                    # y añadirla si necesitamos más
                    img = np.expand_dims(img_original, 0) # Añadir dimensión de batch
                    
                    # Usamos .flow para generar 1 imagen aumentada
                    # next() porque .flow devuelve un generador
                    augmented_batch = next(datagen.flow(img, batch_size=1))
                    augmented_images_for_class.append(augmented_batch[0])

            # Añadir las imágenes aumentadas al conjunto total
            X_balanced.extend(augmented_images_for_class)
            y_balanced.extend([class_idx] * len(augmented_images_for_class))


    num_final_classes = len(class_names_list)
    print(f"Total de imágenes balanceadas: {len(X_balanced)}")

    final_counts = Counter(y_balanced)
    print("Distribución final de clases después del balanceo:")
    for cls in range(len(class_names_list)):
        if cls in final_counts:
            print(f"    Clase {class_names_list[cls]}: {final_counts[cls]}")

    return np.array(X_balanced), tf.keras.utils.to_categorical(np.array(y_balanced), num_classes=num_final_classes)


def create_model_with_efficientnetb0(input_shape, num_classes_model, learning_rate=0.001, fine_tuning_lr=1e-5):
    """
    Crea un modelo basado en EfficientNetB0 con capas densas personalizadas.
    """
    base_model = EfficientNetB0(weights='imagenet', include_top=False, input_shape=input_shape)

    # Congelar capas de la base del modelo por defecto
    base_model.trainable = False

    x = base_model.output
    x = GlobalAveragePooling2D()(x)

    x = Dense(512, activation='relu', kernel_regularizer=regularizers.l2(0.001))(x)
    x = BatchNormalization()(x)
    x = Dropout(0.3)(x)
    x = Dense(256, activation='relu', kernel_regularizer=regularizers.l2(0.001))(x)
    x = Dropout(0.2)(x)
    outputs = Dense(num_classes_model, activation='softmax')(x)

    model = Model(inputs=base_model.input, outputs=outputs)

    # Compilar para la fase 1 (capas densas)
    model.compile(optimizer=Adam(learning_rate=learning_rate),
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

    return model, base_model

def plot_incorrect_predictions(incorrect_preds, class_names_list, img_size_display, num_to_plot=40, cols=5, title_suffix=""):
    """
    Visualiza una selección de imágenes que el modelo predijo incorrectamente,
    incluyendo la confianza de la predicción incorrecta.
    """
    if not incorrect_preds:
        print("No hay predicciones incorrectas para mostrar.")
        return

    if len(incorrect_preds) > num_to_plot:
        print(f"\nMostrando {num_to_plot} de {len(incorrect_preds)} predicciones incorrectas al azar{title_suffix}...")
        preds_to_plot = random.sample(incorrect_preds, num_to_plot)
    else:
        print(f"\nMostrando todas las {len(incorrect_preds)} predicciones incorrectas{title_suffix}...")
        preds_to_plot = incorrect_preds

    rows = (len(preds_to_plot) + cols - 1) // cols
    plt.figure(figsize=(cols * 4, rows * 4))

    for i, pred_info in enumerate(preds_to_plot):
        plt.subplot(rows, cols, i + 1)
        try:
            img = tf.keras.preprocessing.image.load_img(pred_info['filepath'], target_size=(img_size_display, img_size_display))
            plt.imshow(img)
            plt.title(f"Real: {pred_info['true_label']}\nPred: {pred_info['pred_label']} ({pred_info['pred_confidence']:.2%})\n{pred_info.get('fold', '')}",
                      fontsize=10, color='red') # 'fold' es opcional
            plt.axis('off')
        except Exception as e:
            print(f"Error al cargar/mostrar imagen {pred_info['filepath']}: {e}")
            plt.title("Error de imagen", color='red')
            plt.axis('off')

    plt.suptitle(f"Imágenes Predichas Incorrectamente {title_suffix}", fontsize=16)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

def analyze_confidence_by_class(true_labels, pred_probs, class_names_list, title_suffix=""):
    """
    Calcula y muestra las estadísticas de confianza (probabilidad máxima)
    para las predicciones correctas e incorrectas, agrupadas por clase.
    """
    print(f"\n--- Análisis de Confianza por Clase {title_suffix} ---")

    true_labels = np.array(true_labels)
    pred_probs = np.array(pred_probs)
    pred_labels = np.argmax(pred_probs, axis=1)

    max_confidences = np.max(pred_probs, axis=1)
    is_correct = (true_labels == pred_labels)

    confidence_data = {
        'Class': [],
        'Correct_Avg_Conf': [],
        'Incorrect_Avg_Conf': [],
        'Overall_Avg_Conf': []
    }

    for i, class_name in enumerate(class_names_list):
        class_indices = np.where(true_labels == i)[0]
        if len(class_indices) == 0:
            print(f"Clase '{class_name}': No hay muestras en el conjunto {title_suffix}.")
            continue

        class_confidences = max_confidences[class_indices]
        class_correct_flags = is_correct[class_indices]

        correct_confidences = class_confidences[class_correct_flags]
        avg_correct_conf = np.mean(correct_confidences) if len(correct_confidences) > 0 else np.nan

        incorrect_confidences = class_confidences[~class_correct_flags]
        avg_incorrect_conf = np.mean(incorrect_confidences) if len(incorrect_confidences) > 0 else np.nan

        overall_avg_conf = np.mean(class_confidences)

        confidence_data['Class'].append(class_name)
        confidence_data['Correct_Avg_Conf'].append(avg_correct_conf)
        confidence_data['Incorrect_Avg_Conf'].append(avg_incorrect_conf)
        confidence_data['Overall_Avg_Conf'].append(overall_avg_conf)

        print(f"\nClase: {class_name}")
        print(f"  Confianza promedio (Correctas): {avg_correct_conf:.4f} (N={len(correct_confidences)})")
        print(f"  Confianza promedio (Incorrectas): {avg_incorrect_conf:.4f} (N={len(incorrect_confidences)})")
        print(f"  Confianza promedio (General): {overall_avg_conf:.4f} (N={len(class_indices)})")

    df_confidence = pd.DataFrame(confidence_data)
    print(f"\nResumen de Confianza por Clase {title_suffix}:")
    print(df_confidence.round(4).to_string(index=False))


# --- Inicio del Pipeline Principal ---

print("Cargando todas las imágenes del dataset...")
X_full, y_full, class_names, all_filepaths = load_all_images(DATASET_PATH, IMG_SIZE)
if X_full.size == 0:
    raise ValueError("No se cargaron imágenes. Verifica DATASET_PATH y el contenido de las carpetas.")
print(f"Total de imágenes cargadas: {len(X_full)}, Clases: {class_names}")
num_classes = len(class_names)

# --- Paso 1: Separar un 10% para el conjunto de prueba (TEST) ---
print(f"\nSeparando {TEST_SPLIT_PERCENTAGE*100:.0f}% ({int(len(X_full) * TEST_SPLIT_PERCENTAGE)}) de imágenes para el conjunto de prueba...")
X_train_val, X_test, y_train_val, y_test, filepaths_train_val, filepaths_test = train_test_split(
    X_full, y_full, all_filepaths,
    test_size=TEST_SPLIT_PERCENTAGE,
    stratify=y_full, # Asegura que haya representación de todas las clases
    random_state=RANDOM_SEED
)

print(f"Imágenes para K-Fold (entrenamiento+validación): {len(X_train_val)}")
print(f"Imágenes para Test final: {len(X_test)}")

# Verificar que haya al menos una imagen de cada clase en el test set
test_class_counts = Counter(y_test)
missing_test_classes = [class_names[i] for i in range(num_classes) if i not in test_class_counts]
if missing_test_classes:
    print(f"Advertencia: Las siguientes clases no están presentes en el conjunto de prueba: {missing_test_classes}. Reajuste TEST_SPLIT_PERCENTAGE o RANDOM_SEED si esto es crítico.")
else:
    print("Todas las clases están representadas en el conjunto de prueba.")
    print("Distribución de clases en el conjunto de prueba:")
    for class_idx in sorted(test_class_counts.keys()):
        print(f"    - {class_names[class_idx]}: {test_class_counts[class_idx]} imágenes")

# --- Paso 2: Ejecutar K-Fold Cross-Validation en el conjunto de Entrenamiento+Validación ---
if USE_STRATIFIED_KFOLD:
    kf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=RANDOM_SEED)
    cv_type = "Estratificada K-Fold Cross-Validation"
    split_iterator = kf.split(X_train_val, y_train_val)
else:
    kf = KFold(n_splits=N_SPLITS, shuffle=True, random_state=RANDOM_SEED)
    cv_type = "K-Fold Cross-Validation"
    split_iterator = kf.split(X_train_val)

all_true_labels_kf = []
all_pred_labels_kf = []
all_pred_probs_kf = []
fold_accuracies_kf = []
fold_histories = []

incorrect_predictions_info_kf = [] # Para K-Fold

print(f"\nIniciando {cv_type} con {N_SPLITS} folds en el conjunto de entrenamiento+validación...")

for fold_idx, (train_indices, val_indices) in enumerate(split_iterator):
    print(f"\n{'='*50}")
    print(f"Fold K-Fold {fold_idx+1}/{N_SPLITS}")
    print(f"{'='*50}")

    X_train_kf, X_val_kf = X_train_val[train_indices], X_train_val[val_indices]
    y_train_kf, y_val_kf = y_train_val[train_indices], y_train_val[val_indices]
    filepaths_val_kf = filepaths_train_val[val_indices]

    y_val_kf_categorical = tf.keras.utils.to_categorical(y_val_kf, num_classes=num_classes)

    X_train_kf_processed, y_train_kf_categorical = balance_dataset_with_augmentation(
        X_train_kf, y_train_kf, class_names, IMG_SIZE
    )

    if X_train_kf_processed.size == 0:
        print(f"Fold {fold_idx+1} no tiene datos de entrenamiento válidos, saltando.")
        fold_accuracies_kf.append(0)
        continue

    model_kf, base_model_kf = create_model_with_efficientnetb0((IMG_SIZE, IMG_SIZE, 3), num_classes)

    early_stopping_phase1 = EarlyStopping(monitor='val_loss', patience=7, verbose=1, restore_best_weights=True)
    reduce_lr_phase1 = ReduceLROnPlateau(monitor='val_loss', factor=0.3, patience=2, min_lr=1e-7, verbose=1)

    # --- Configuración del ModelCheckpoint para cada fold (SIEMPRE se crea) ---
    os.makedirs(f'/kaggle/working/kfold_models/fold_{fold_idx+1}/', exist_ok=True)
    checkpoint_filepath_fold = f'/kaggle/working/kfold_models/fold_{fold_idx+1}/best_model_fold_{fold_idx+1}.h5'
    model_checkpoint_fold = ModelCheckpoint(
        filepath=checkpoint_filepath_fold,
        monitor='val_accuracy', # Monitorea la precisión de validación
        save_best_only=True,    # Solo guarda el mejor modelo
        mode='max',             # Se busca el valor máximo
        verbose=1
    )
    # Lista de callbacks para la Fase 1
    callbacks_phase1 = [early_stopping_phase1, reduce_lr_phase1, model_checkpoint_fold]


    print("\nFASE 1: Entrenamiento de capas densas con EfficientNetB0 congelado")
    history_phase1 = model_kf.fit(
        X_train_kf_processed, y_train_kf_categorical,
        epochs=50, # Épocas iniciales para capas densas
        batch_size=BATCH_SIZE,
        validation_data=(X_val_kf, y_val_kf_categorical),
        callbacks=callbacks_phase1, # Incluye el checkpoint aquí
        verbose=1
    )

    epochs_phase1 = len(history_phase1.history['loss'])
    print(f"Fase 1 completada después de {epochs_phase1} épocas")

    if PERFORM_FINE_TUNING:
        print("\nFASE 2: Fine-tuning de EfficientNetB0 (activado)")
        # Descongelar las últimas 10 capas del modelo base para fine-tuning
        for layer in base_model_kf.layers[-10:]:
            layer.trainable = True

        model_kf.compile(
            optimizer=Adam(learning_rate=1e-5), # Tasa de aprendizaje más pequeña para fine-tuning
            loss='categorical_crossentropy',
            metrics=['accuracy']
        )

        early_stopping_phase2 = EarlyStopping(monitor='val_loss', patience=7, verbose=1, restore_best_weights=True)
        reduce_lr_phase2 = ReduceLROnPlateau(monitor='val_loss', factor=0.4, patience=2, min_lr=1e-7, verbose=1)
        
        # Lista de callbacks para la Fase 2 (el checkpoint ya está monitoreando)
        callbacks_phase2 = [early_stopping_phase2, reduce_lr_phase2, model_checkpoint_fold]

        history_phase2 = model_kf.fit(
            X_train_kf_processed, y_train_kf_categorical,
            epochs=EPOCHS_KFOLD, # Épocas totales, empezando desde epochs_phase1
            initial_epoch=epochs_phase1,
            batch_size=BATCH_SIZE,
            validation_data=(X_val_kf, y_val_kf_categorical),
            callbacks=callbacks_phase2, 
            verbose=1
        )

        combined_history = {}
        combined_history.update(history_phase1.history)
        for key in history_phase2.history:
            if key in combined_history:
                combined_history[key].extend(history_phase2.history[key])
            else:
                combined_history[key] = history_phase2.history[key]
    else:
        print("\nFASE 2: Fine-tuning de EfficientNetB0 (desactivado). El modelo se entrenó solo con las capas densas.")
        combined_history = history_phase1.history

    fold_histories.append(combined_history)

    # Cargar el mejor modelo guardado por el checkpoint para hacer las predicciones del fold
    # Esto asegura que evaluamos el modelo que tuvo el mejor rendimiento en validación para este fold
    if os.path.exists(checkpoint_filepath_fold):
        model_kf_best = tf.keras.models.load_model(checkpoint_filepath_fold)
        print(f"Cargado el mejor modelo para Fold {fold_idx+1} desde {checkpoint_filepath_fold}")
    else:
        # Si por alguna razón no se guardó (ej. entrenamiento muy corto o problema), usa el último entrenado
        print(f"Advertencia: No se encontró el mejor modelo guardado para Fold {fold_idx+1}. Usando el modelo actual en memoria.")
        model_kf_best = model_kf

    y_pred_prob_fold = model_kf_best.predict(X_val_kf, verbose=0)
    y_pred_class_fold = np.argmax(y_pred_prob_fold, axis=1)

    all_true_labels_kf.extend(y_val_kf)
    all_pred_labels_kf.extend(y_pred_class_fold)
    all_pred_probs_kf.extend(y_pred_prob_fold)

    accuracy_this_fold = accuracy_score(y_val_kf, y_pred_class_fold)
    fold_accuracies_kf.append(accuracy_this_fold)

    print(f"Fold {fold_idx+1}: Accuracy de validación: {accuracy_this_fold*100:.2f}%")

    # Almacenar la confianza de la predicción incorrecta para K-Fold
    for i in range(len(y_val_kf)):
        if y_val_kf[i] != y_pred_class_fold[i]:
            incorrect_predictions_info_kf.append({
                'filepath': filepaths_val_kf[i],
                'true_label': class_names[y_val_kf[i]],
                'pred_label': class_names[y_pred_class_fold[i]],
                'pred_confidence': y_pred_prob_fold[i, y_pred_class_fold[i]],
                'fold': f"Fold {fold_idx + 1}" 
            })

    del model_kf
    del model_kf_best 
    tf.keras.backend.clear_session()
    import gc
    gc.collect()

if not all_true_labels_kf:
    print("No se generaron predicciones válidas en K-Fold Cross-Validation.")
else:
    mean_kf_accuracy = np.mean(fold_accuracies_kf)
    std_kf_accuracy = np.std(fold_accuracies_kf)

    print(f"\n{'='*70}")
    print(f"RESULTADOS DE {cv_type} (sobre el 90% de datos para Entrenamiento+Validación)")
    print(f"{'='*70}")
    print(f"Precisión Promedio: {mean_kf_accuracy*100:.2f}% ± {std_kf_accuracy*100:.2f}%")
    print(f"Precisión por fold: {[f'{acc*100:.2f}%' for acc in fold_accuracies_kf]}")

    conf_matrix_kf = confusion_matrix(all_true_labels_kf, all_pred_labels_kf, labels=list(range(num_classes)))
    plt.figure(figsize=(12, 10))
    sns.heatmap(conf_matrix_kf, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names)
    plt.title(f'Matriz de Confusión Global - {cv_type}')
    plt.xlabel('Predicción')
    plt.ylabel('Real')
    plt.tight_layout()
    plt.show()

    row_sums = conf_matrix_kf.sum(axis=1, keepdims=True)
    norm_conf_matrix = conf_matrix_kf / row_sums

    plt.figure(figsize=(12, 10))
    sns.heatmap(norm_conf_matrix, annot=True, fmt='.2f', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names)
    plt.title(f'Matriz de Confusión Normalizada - {cv_type}')
    plt.xlabel('Predicción')
    plt.ylabel('Real')
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    for i, history in enumerate(fold_histories):
        plt.plot(history['accuracy'], label=f'Fold {i+1} train', alpha=0.7)
        plt.plot(history['val_accuracy'], label=f'Fold {i+1} val', alpha=0.7, linestyle='--')

    plt.title('Precisión durante entrenamiento (K-Fold)')
    plt.xlabel('Época')
    plt.ylabel('Precisión')
    plt.legend()

    plt.subplot(1, 2, 2)
    for i, history in enumerate(fold_histories):
        plt.plot(history['loss'], label=f'Fold {i+1} train', alpha=0.7)
        plt.plot(history['val_loss'], label=f'Fold {i+1} val', alpha=0.7, linestyle='--')

    plt.title('Pérdida durante entrenamiento (K-Fold)')
    plt.xlabel('Época')
    plt.ylabel('Pérdida')
    plt.legend()

    plt.tight_layout()
    plt.show()

    print(f"\nInforme de Clasificación - {cv_type}:")
    print(classification_report(all_true_labels_kf, all_pred_labels_kf,
                                labels=list(range(num_classes)),
                                target_names=class_names,
                                zero_division=0))

    all_pred_probs_kf_arr = np.array(all_pred_probs_kf)
    max_probs_global = np.max(all_pred_probs_kf_arr, axis=1)
    correct_predictions_global = np.array(all_true_labels_kf) == np.array(all_pred_labels_kf)

    plt.figure(figsize=(10, 6))
    plt.hist([max_probs_global[correct_predictions_global], max_probs_global[~correct_predictions_global]],
             bins=20, alpha=0.7, label=['Correctas', 'Incorrectas'], color=['green', 'red'])
    plt.xlabel('Confianza de Predicción (Probabilidad Máxima)')
    plt.ylabel('Número de Predicciones')
    plt.title('Distribución Global de Confianza de Predicción (K-Fold)')
    plt.legend()
    plt.tight_layout()
    plt.show()

    print("\nEstadísticas de Confianza Global (K-Fold):")
    print(f"Confianza promedio en predicciones correctas: {max_probs_global[correct_predictions_global].mean():.4f}")
    print(f"Confianza promedio en predicciones incorrectas: {max_probs_global[~correct_predictions_global].mean():.4f}")
    print(f"Número de predicciones con confianza >0.9: {np.sum(max_probs_global > 0.9)}")
    print(f"Precisión en predicciones con confianza >0.9: {np.mean(correct_predictions_global[max_probs_global > 0.9]):.4f}")

    analyze_confidence_by_class(all_true_labels_kf, all_pred_probs_kf, class_names, title_suffix="(K-Fold)")

    if incorrect_predictions_info_kf:
        plot_incorrect_predictions(incorrect_predictions_info_kf, class_names, IMG_SIZE, num_to_plot=40, cols=5, title_suffix="(K-Fold)")


# --- Paso 3: Entrenamiento del Modelo Final con TODOS los datos de Entrenamiento+Validación ---
print(f"\n{'='*70}")
print("ENTRENAMIENTO DEL MODELO FINAL CON EL 90% DE LOS DATOS (ENTRENAMIENTO + VALIDACIÓN)")
print(f"{'='*70}")

# Convertir y_train_val a formato categórico para el entrenamiento final
y_train_val_categorical_final = tf.keras.utils.to_categorical(y_train_val, num_classes=num_classes)

# Balancear todo el conjunto de entrenamiento+validación para el entrenamiento final
X_train_val_final_processed, y_train_val_final_categorical = balance_dataset_with_augmentation(
    X_train_val, y_train_val, class_names, IMG_SIZE
)

model_final, base_model_final = create_model_with_efficientnetb0((IMG_SIZE, IMG_SIZE, 3), num_classes)

early_stopping_final_phase1 = EarlyStopping(monitor='val_loss', patience=7, verbose=1, restore_best_weights=True)
reduce_lr_final_phase1 = ReduceLROnPlateau(monitor='val_loss', factor=0.3, patience=2, min_lr=1e-7, verbose=1)

# --- Configuración del ModelCheckpoint para el modelo final  ---
os.makedirs('/kaggle/working/final_model/', exist_ok=True)
final_model_filepath = '/kaggle/working/final_model/final_efficientnetb0_model.h5'
model_checkpoint_final = ModelCheckpoint(
    filepath=final_model_filepath,
    monitor='val_accuracy', 
    save_best_only=True,    
    mode='max',             
    verbose=1
)
# Lista de callbacks para la Fase 1 del modelo final
callbacks_final_phase1 = [early_stopping_final_phase1, reduce_lr_final_phase1, model_checkpoint_final]


print("\nFASE 1: Entrenamiento de capas densas del modelo final con EfficientNetB0 congelado")
history_final_phase1 = model_final.fit(
    X_train_val_final_processed, y_train_val_final_categorical,
    epochs=20, # Épocas iniciales para capas densas
    batch_size=BATCH_SIZE,
    validation_split=0.15, 
    callbacks=callbacks_final_phase1, 
    verbose=1
)

epochs_final_phase1 = len(history_final_phase1.history['loss'])
print(f"Fase 1 del modelo final completada después de {epochs_final_phase1} épocas")

if PERFORM_FINE_TUNING:
    print("\nFASE 2: Fine-tuning de EfficientNetB0 para el modelo final (activado)")
    for layer in base_model_final.layers[-10:]: 
        layer.trainable = True

    model_final.compile(
        optimizer=Adam(learning_rate=1e-5),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    early_stopping_final_phase2 = EarlyStopping(monitor='val_loss', patience=7, verbose=1, restore_best_weights=True)
    reduce_lr_final_phase2 = ReduceLROnPlateau(monitor='val_loss', factor=0.4, patience=2, min_lr=1e-7, verbose=1)
    
    # Lista de callbacks para la Fase 2 del modelo final
    callbacks_final_phase2 = [early_stopping_final_phase2, reduce_lr_final_phase2, model_checkpoint_final]

    history_final_phase2 = model_final.fit(
        X_train_val_final_processed, y_train_val_final_categorical,
        epochs=EPOCHS_KFOLD, 
        initial_epoch=epochs_final_phase1,
        batch_size=BATCH_SIZE,
        validation_split=0.15,
        callbacks=callbacks_final_phase2, 
        verbose=1
    )
else:
    print("\nFASE 2: Fine-tuning de EfficientNetB0 para el modelo final (desactivado).")

print(f"Modelo final entrenado con {len(X_train_val_final_processed)} imágenes (incluyendo aumentaciones).")

# --- Paso 4: Evaluación del Modelo Final en el Conjunto de Prueba ---
print(f"\n{'='*70}")
print("EVALUACIÓN DEL MODELO FINAL EN EL CONJUNTO DE PRUEBA")
print(f"{'='*70}")

y_test_categorical = tf.keras.utils.to_categorical(y_test, num_classes=num_classes)

# Cargar el mejor modelo guardado por el checkpoint
if os.path.exists(final_model_filepath):
    print(f"Cargando el mejor modelo final guardado desde: {final_model_filepath}")
    final_model_for_evaluation = tf.keras.models.load_model(final_model_filepath)
else:
    # Si por alguna razón no se guardó (ej. error inesperado, entrenamiento muy corto sin mejoras), usa el último entrenado
    print("Advertencia: No se encontró el mejor modelo final guardado. Usando el modelo actual en memoria.")
    final_model_for_evaluation = model_final

test_loss, test_accuracy = final_model_for_evaluation.evaluate(X_test, y_test_categorical, verbose=0)
print(f"\nPrecisión del modelo final en el conjunto de prueba: {test_accuracy*100:.2f}%")

y_pred_prob_test = final_model_for_evaluation.predict(X_test, verbose=0)
y_pred_class_test = np.argmax(y_pred_prob_test, axis=1)

# Calcular F1-score
f1_test_weighted = f1_score(y_test, y_pred_class_test, average='weighted')
f1_test_macro = f1_score(y_test, y_pred_class_test, average='macro')
print(f"F1-score ponderado del modelo final en el conjunto de prueba: {f1_test_weighted:.4f}")
print(f"F1-score macro del modelo final en el conjunto de prueba: {f1_test_macro:.4f}")

# Matriz de Confusión para el conjunto de prueba
conf_matrix_test = confusion_matrix(y_test, y_pred_class_test, labels=list(range(num_classes)))
plt.figure(figsize=(12, 10))
sns.heatmap(conf_matrix_test, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title('Matriz de Confusión - Conjunto de Prueba')
plt.xlabel('Predicción')
plt.ylabel('Real')
plt.tight_layout()
plt.show()

# Matriz de Confusión Normalizada para el conjunto de prueba
row_sums_test = conf_matrix_test.sum(axis=1, keepdims=True)
norm_conf_matrix_test = conf_matrix_test / row_sums_test

plt.figure(figsize=(12, 10))
sns.heatmap(norm_conf_matrix_test, annot=True, fmt='.2f', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title('Matriz de Confusión Normalizada - Conjunto de Prueba')
plt.xlabel('Predicción')
plt.ylabel('Real')
plt.tight_layout()
plt.show()

print("\nInforme de Clasificación - Conjunto de Prueba:")
print(classification_report(y_test, y_pred_class_test,
                            labels=list(range(num_classes)),
                            target_names=class_names,
                            zero_division=0))

# Estadísticas de Confianza para el conjunto de prueba
all_pred_probs_test_arr = np.array(y_pred_prob_test)
max_probs_test = np.max(all_pred_probs_test_arr, axis=1)
correct_predictions_test = np.array(y_test) == np.array(y_pred_class_test)

plt.figure(figsize=(10, 6))
plt.hist([max_probs_test[correct_predictions_test], max_probs_test[~correct_predictions_test]],
         bins=20, alpha=0.7, label=['Correctas', 'Incorrectas'], color=['green', 'red'])
plt.xlabel('Confianza de Predicción (Probabilidad Máxima)')
plt.ylabel('Número de Predicciones')
plt.title('Distribución Global de Confianza de Predicción (Conjunto de Prueba)')
plt.legend()
plt.tight_layout()
plt.show()

print("\nEstadísticas de Confianza Global (Conjunto de Prueba):")
print(f"Confianza promedio en predicciones correctas: {max_probs_test[correct_predictions_test].mean():.4f}")
print(f"Confianza promedio en predicciones incorrectas: {max_probs_test[~correct_predictions_test].mean():.4f}")
print(f"Número de predicciones con confianza >0.9: {np.sum(max_probs_test > 0.9)}")
print(f"Precisión en predicciones con confianza >0.9: {np.mean(correct_predictions_test[max_probs_test > 0.9]):.4f}")

analyze_confidence_by_class(y_test, y_pred_prob_test, class_names, title_suffix="(Conjunto de Prueba)")

# Visualización de predicciones incorrectas en el conjunto de prueba
incorrect_predictions_info_test = []
for i in range(len(y_test)):
    if y_test[i] != y_pred_class_test[i]:
        incorrect_predictions_info_test.append({
            'filepath': filepaths_test[i],
            'true_label': class_names[y_test[i]],
            'pred_label': class_names[y_pred_class_test[i]],
            'pred_confidence': y_pred_prob_test[i, y_pred_class_test[i]],
            'fold': "Test Set" # Identificador para el conjunto de prueba
        })

if incorrect_predictions_info_test:
    plot_incorrect_predictions(incorrect_predictions_info_test, class_names, IMG_SIZE, num_to_plot=40, cols=5, title_suffix="(Conjunto de Prueba)")

print("\nScript de evaluación completo finalizado.")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.applications.efficientnet import preprocess_input
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, f1_score
from collections import Counter
import random

# --- Configuración General ---

TEST_DATASET_PATH = '/kaggle/input/psoriasis-ds/BaseDatos1/BaseDatos1/test'
IMG_SIZE = 224
BATCH_SIZE = 16
MODEL_PATH = '/kaggle/working/final_model/final_efficientnetb0_model.h5'

EXPECTED_CLASS_NAMES = [
    'Psoriasis_Eritrodermica',
    'Psoriasis_Guttata',
    'Psoriasis_Inversa',
    'Psoriasis_Pustulosa',
    'Psoriasis_Vulgar'
]

# Creamos una lista de "palabras clave" más flexibles para buscar en los nombres de archivo
# Estas son las partes importantes de los nombres de clase que esperamos encontrar
CLASS_KEYWORDS = {
    'erythrodermic': 'Psoriasis_Eritrodermica',
    'guttate': 'Psoriasis_Guttata',
    'sunburn': 'Psoriasis Guttata',
    'inversus': 'Psoriasis_Inversa',
    'penis': 'Psoriasis_Inversa',
    'pustular': 'Psoriasis_Pustulosa', 
    'palms_soles': 'Psoriasis_Pustulosa',
    'nails': 'Psoriasis_Pustulosa',
    'vulgar': 'Psoriasis_Vulgar',
    'scalp': 'Psoriasis_Vulgar',
    'plaque': 'Psoriasis_Vulgar',
}


# --- Funciones de Preprocesamiento y Evaluación (adaptadas para etiquetas en el nombre del archivo) ---

def load_images_from_flat_directory(directory, img_size, expected_class_names, class_keywords_mapping):
    """
    Carga imágenes de un directorio plano y extrae la etiqueta del nombre del archivo
    usando palabras clave más flexibles. Las imágenes se preprocesan.
    Retorna arrays de imágenes, etiquetas (numéricas) y rutas de archivo originales.
    """
    images, labels, original_filepaths = [], [], []
    # Mapeo de nombre de clase a índice numérico (orden de expected_class_names)
    class_indices = {class_name: i for i, class_name in enumerate(expected_class_names)}

    print(f"Cargando y redimensionando imágenes a {img_size}x{img_size} del directorio: {directory}...")
    
    # Comprobación de que el directorio existe
    if not os.path.exists(directory):
        print(f"Error: El directorio '{directory}' no existe. Por favor, verifica la ruta.")
        return np.array([]), np.array([]), expected_class_names, np.array([])
    
    for img_name in os.listdir(directory):
        if img_name.lower().endswith(('.png', '.jpg', '.jpeg')):
            img_path = os.path.join(directory, img_name)
            
            # Normalizar el nombre del archivo para la búsqueda (todo a minúsculas y reemplazar guiones)
            normalized_img_name = img_name.lower().replace('-', '_')
            
            found_class_name = None
            # Iterar a través de las palabras clave para encontrar una coincidencia
            for keyword, mapped_class in class_keywords_mapping.items():
                if keyword.lower() in normalized_img_name:
                    found_class_name = mapped_class
                    break
            
            # Si no se encontró una clase principal, intentar con los nombres de clase completos
            if found_class_name is None:
                 for class_name in expected_class_names:
                    # Normalizar también el nombre de clase para la comparación
                    normalized_class_name = class_name.lower().replace('-', '_')
                    if normalized_class_name in normalized_img_name:
                        found_class_name = class_name
                        break
            
            if found_class_name is None:
                print(f"Advertencia: No se pudo determinar una clase principal válida para la imagen '{img_name}'. Saltando.")
                continue

            try:
                img = tf.keras.preprocessing.image.load_img(img_path, target_size=(img_size, img_size))
                img_array = tf.keras.preprocessing.image.img_to_array(img)
                img_array = preprocess_input(img_array)
                images.append(img_array)
                # Asegurarse de que la clase encontrada está en el mapeo de class_indices
                if found_class_name in class_indices:
                    labels.append(class_indices[found_class_name])
                    original_filepaths.append(img_path)
                else:
                    print(f"Advertencia: Clase '{found_class_name}' extraída de '{img_name}' no está en EXPECTED_CLASS_NAMES. Saltando.")
            except Exception as e:
                print(f"Error cargando imagen {img_path}: {e}")

    print(f"Total de imágenes cargadas y clasificadas del nuevo dataset: {len(images)}")
    return np.array(images), np.array(labels), expected_class_names, np.array(original_filepaths)

def plot_incorrect_predictions(incorrect_preds, class_names_list, img_size_display, num_to_plot=20, cols=4, title_suffix=""):
    """
    Visualiza una selección de imágenes que el modelo predijo incorrectamente,
    incluyendo la confianza de la predicción incorrecta.
    """
    if not incorrect_preds:
        print("No hay predicciones incorrectas para mostrar.")
        return

    if len(incorrect_preds) > num_to_plot:
        print(f"\nMostrando {num_to_plot} de {len(incorrect_preds)} predicciones incorrectas al azar{title_suffix}...")
        preds_to_plot = random.sample(incorrect_preds, num_to_plot)
    else:
        print(f"\nMostrando todas las {len(incorrect_preds)} predicciones incorrectas{title_suffix}...")
        preds_to_plot = incorrect_preds

    rows = (len(preds_to_plot) + cols - 1) // cols
    plt.figure(figsize=(cols * 4, rows * 4))

    for i, pred_info in enumerate(preds_to_plot):
        plt.subplot(rows, cols, i + 1)
        try:
            img = tf.keras.preprocessing.image.load_img(pred_info['filepath'], target_size=(img_size_display, img_size_display))
            plt.imshow(img)
            plt.title(f"Real: {pred_info['true_label']}\nPred: {pred_info['pred_label']} ({pred_info['pred_confidence']:.2%})",
                      fontsize=10, color='red')
            plt.axis('off')
        except Exception as e:
            print(f"Error al cargar/mostrar imagen {pred_info['filepath']}: {e}")
            plt.title("Error de imagen", color='red')
            plt.axis('off')

    plt.suptitle(f"Imágenes Predichas Incorrectamente {title_suffix}", fontsize=16)
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

def analyze_confidence_by_class(true_labels, pred_probs, class_names_list, title_suffix=""):
    """
    Calcula y muestra las estadísticas de confianza (probabilidad máxima)
    para las predicciones correctas e incorrectas, agrupadas por clase.
    """
    print(f"\n--- Análisis de Confianza por Clase {title_suffix} ---")

    true_labels = np.array(true_labels)
    pred_probs = np.array(pred_probs)
    pred_labels = np.argmax(pred_probs, axis=1)

    max_confidences = np.max(pred_probs, axis=1)
    is_correct = (true_labels == pred_labels)

    confidence_data = {
        'Class': [],
        'Correct_Avg_Conf': [],
        'Incorrect_Avg_Conf': [],
        'Overall_Avg_Conf': []
    }

    for i, class_name in enumerate(class_names_list):
        class_indices = np.where(true_labels == i)[0]
        if len(class_indices) == 0:
            print(f"Clase '{class_name}': No hay muestras en el conjunto {title_suffix}.")
            continue

        class_confidences = max_confidences[class_indices]
        class_correct_flags = is_correct[class_indices]

        correct_confidences = class_confidences[class_correct_flags]
        avg_correct_conf = np.mean(correct_confidences) if len(correct_confidences) > 0 else np.nan

        incorrect_confidences = class_confidences[~class_correct_flags]
        avg_incorrect_conf = np.mean(incorrect_confidences) if len(incorrect_confidences) > 0 else np.nan

        overall_avg_conf = np.mean(class_confidences)

        confidence_data['Class'].append(class_name)
        confidence_data['Correct_Avg_Conf'].append(avg_correct_conf)
        confidence_data['Incorrect_Avg_Conf'].append(avg_incorrect_conf)
        confidence_data['Overall_Avg_Conf'].append(overall_avg_conf)

        print(f"\nClase: {class_name}")
        print(f"  Confianza promedio (Correctas): {avg_correct_conf:.4f} (N={len(correct_confidences)})")
        print(f"  Confianza promedio (Incorrectas): {avg_incorrect_conf:.4f} (N={len(incorrect_confidences)})")
        print(f"  Confianza promedio (General): {overall_avg_conf:.4f} (N={len(class_indices)})")

    df_confidence = pd.DataFrame(confidence_data)
    print(f"\nResumen de Confianza por Clase {title_suffix}:")
    print(df_confidence.round(4).to_string(index=False))



### Pipeline de Evaluación del Nuevo Dataset (Dataset 1)

# --- Pipeline de Evaluación del Nuevo Dataset ---

print(f"Cargando el nuevo dataset de prueba desde: {TEST_DATASET_PATH}")
# Usamos la nueva función para directorios planos y pasamos las palabras clave
X_new_test, y_new_test, new_class_names, filepaths_new_test = load_images_from_flat_directory(
    TEST_DATASET_PATH, IMG_SIZE, EXPECTED_CLASS_NAMES, CLASS_KEYWORDS
)

# Comprobación de que se cargaron imágenes
if X_new_test.size == 0:
    raise ValueError("No se cargaron imágenes del nuevo dataset. Verifica TEST_DATASET_PATH y que los nombres de los archivos contengan las etiquetas de clase válidas para tu modelo.")

print(f"\nNúmero de clases detectadas en el nuevo dataset: {len(new_class_names)}")
print(f"Clases detectadas en el nuevo dataset: {new_class_names}")

# Cargar el modelo final entrenado
print(f"\nCargando el modelo entrenado desde: {MODEL_PATH}")
try:
    model = tf.keras.models.load_model(MODEL_PATH)
    print("Modelo cargado exitosamente.")
except Exception as e:
    raise RuntimeError(f"Error al cargar el modelo desde {MODEL_PATH}. Asegúrate de que la ruta es correcta y el archivo existe. Error: {e}")

# Asegurarse de que las etiquetas del nuevo dataset estén en formato categórico para evaluación
y_new_test_categorical = tf.keras.utils.to_categorical(y_new_test, num_classes=len(new_class_names))

# Evaluación del modelo en el nuevo dataset
print(f"\n{'='*70}")
print("EVALUACIÓN DEL MODELO EN EL NUEVO CONJUNTO DE PRUEBA (Dataset 1)")
print(f"{'='*70}")

loss, accuracy = model.evaluate(X_new_test, y_new_test_categorical, verbose=0)
print(f"\nPrecisión del modelo en el nuevo conjunto de prueba: {accuracy*100:.2f}%")

y_pred_prob = model.predict(X_new_test, verbose=0)
y_pred_class = np.argmax(y_pred_prob, axis=1)

# Calcular F1-score
f1_weighted = f1_score(y_new_test, y_pred_class, average='weighted', zero_division=0)
f1_macro = f1_score(y_new_test, y_pred_class, average='macro', zero_division=0)
print(f"F1-score ponderado en el nuevo conjunto de prueba: {f1_weighted:.4f}")
print(f"F1-score macro en el nuevo conjunto de prueba: {f1_macro:.4f}")

# --- Matriz de Confusión ---
conf_matrix = confusion_matrix(y_new_test, y_pred_class, labels=list(range(len(new_class_names))))
plt.figure(figsize=(10, 8))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=new_class_names, yticklabels=new_class_names)
plt.title('Matriz de Confusión - Nuevo Conjunto de Prueba')
plt.xlabel('Predicción')
plt.ylabel('Real')
plt.tight_layout()
plt.show()

# --- Matriz de Confusión Normalizada ---
row_sums = conf_matrix.sum(axis=1, keepdims=True)
norm_conf_matrix = np.where(row_sums == 0, 0, conf_matrix / row_sums) # Evitar división por cero

plt.figure(figsize=(10, 8))
sns.heatmap(norm_conf_matrix, annot=True, fmt='.2f', cmap='Blues',
            xticklabels=new_class_names, yticklabels=new_class_names)
plt.title('Matriz de Confusión Normalizada - Nuevo Conjunto de Prueba')
plt.xlabel('Predicción')
plt.ylabel('Real')
plt.tight_layout()
plt.show()

# --- Informe de Clasificación ---
print("\nInforme de Clasificación - Nuevo Conjunto de Prueba:")
print(classification_report(y_new_test, y_pred_class,
                            labels=list(range(len(new_class_names))),
                            target_names=new_class_names,
                            zero_division=0))

# --- Estadísticas de Confianza Global ---
max_probs = np.max(y_pred_prob, axis=1)
correct_predictions = (y_new_test == y_pred_class)

plt.figure(figsize=(10, 6))
plt.hist([max_probs[correct_predictions], max_probs[~correct_predictions]],
         bins=20, alpha=0.7, label=['Correctas', 'Incorrectas'], color=['green', 'red'])
plt.xlabel('Confianza de Predicción (Probabilidad Máxima)')
plt.ylabel('Número de Predicciones')
plt.title('Distribución de Confianza de Predicción (Nuevo Conjunto de Prueba)')
plt.legend()
plt.tight_layout()
plt.show()

print("\nEstadísticas de Confianza Global (Nuevo Conjunto de Prueba):")
print(f"Confianza promedio en predicciones correctas: {max_probs[correct_predictions].mean():.4f}")
print(f"Confianza promedio en predicciones incorrectas: {max_probs[~correct_predictions].mean():.4f}")
print(f"Número de predicciones con confianza >0.9: {np.sum(max_probs > 0.9)}")
if np.sum(max_probs > 0.9) > 0:
    print(f"Precisión en predicciones con confianza >0.9: {np.mean(correct_predictions[max_probs > 0.9]):.4f}")
else:
    print("No hay predicciones con confianza >0.9 en este conjunto.")

# --- Visualización de Predicciones Incorrectas ---
incorrect_predictions_info = []
for i in range(len(y_new_test)):
    if y_new_test[i] != y_pred_class[i]:
        incorrect_predictions_info.append({
            'filepath': filepaths_new_test[i],
            'true_label': new_class_names[y_new_test[i]],
            'pred_label': new_class_names[y_pred_class[i]],
            'pred_confidence': y_pred_prob[i, y_pred_class[i]]
        })

plot_incorrect_predictions(incorrect_predictions_info, new_class_names, IMG_SIZE, num_to_plot=20, cols=4, title_suffix="(Nuevo Conjunto de Prueba)")

print("\nEvaluación del nuevo dataset completada.")
```
